# 09 · End-to-end demo

Two modes:

1. **Transcript-only** (works offline, fast) — synthetic sample included.
2. **Audio** → Whisper → everything (CPU works; GPU is faster).

CLI equivalents:

```bash
python scripts/generate_minutes.py --input transcript.txt --input-type transcript \
    --output outputs/minutes/demo.md
python scripts/generate_minutes.py --input meeting.wav --output outputs/minutes/demo.md
```

Output is always a **draft for human review**; unknown fields show `Not specified`.

````python
print('[1/6] Choose mode')
import sys; sys.path.insert(0, '.')
MODE = 'transcript'   # 'transcript' or 'audio'
AUDIO = None           # e.g. 'data/samples/sample_meeting.wav' when MODE == 'audio'
print('mode:', MODE)
````

````python
print('[2/6] Provide input')
from pathlib import Path
if MODE == 'transcript':
    INPUT = 'data/samples/sample_meeting.txt'   # or upload your own .txt/.json
    if not Path(INPUT).exists():
        INPUT = None
        try:
            from google.colab import files
            up = files.upload()
            INPUT = list(up)[0]
        except Exception:
            print('No transcript found.')
else:
    INPUT = AUDIO
    if INPUT is None:
        try:
            from google.colab import files
            up = files.upload()
            INPUT = list(up)[0]
        except Exception:
            print('No audio found.')
print('input:', INPUT)
````

````python
print('[3/6] Run the full pipeline')
!python -u scripts/generate_minutes.py --input $INPUT --output outputs/minutes/demo.md --title "Demo Meeting" --date 2026-10-01
````

````python
print('[4/6] Show the generated minutes')
from pathlib import Path
p = Path('outputs/minutes/demo.md')
print(p.read_text(encoding='utf-8') if p.exists() else 'no output yet')
````

````python
print('[5/6] Audit sidecar (JSON) — every item keeps its source')
import json
side = Path('outputs/minutes/demo.json')
if side.exists():
    data = json.loads(side.read_text(encoding='utf-8'))
    print('decisions:', len(data['decisions']), '| actions:', len(data['actions']))
    print('review status:', data['review_status'])
    for a in data['actions']:
        print(' -', a['action'], '| owner:', a['owner'], '| deadline:', a['deadline'])
````
